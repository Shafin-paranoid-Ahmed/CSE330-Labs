```python
#Question 2 a
def x_vondo(data_x):
    n_nodes = len(data_x)
    v = np.array((n_nodes,n_nodes))
    for i in range(n_nodes):
        for j in range(n_nodes):
            v[i][j] = data_x[i]**j
    v_inv = np.linalgoinv(v)
    return v_inv
data_x = np.array([1,2,3,4])
print(x_vondo(data_x))
```
